```python
df = pd.read_csv("https://raw.githubusercontent.com/DataTalksClub/machine-learning-zoomcamp/main/cohorts/2026/data/car_fuel_efficiency_2026.csv")
```

```python
print(df.shape)
print(df.columns)
print(df.head())
print(df.isnull().any().sum())
print(df[df["origin"] == "Asia"]["fuel_efficiency_mpg"].max())
```

```python


print(df["horsepower"].median())

mode_horsepower = df["horsepower"].mode()
mode_horsepower = mode_horsepower.iloc[0]
fill_df = df["horsepower"].fillna(mode_horsepower)

print(fill_df.median())
```

```python
X = df[df["origin"] == 'Asia'][["vehicle_weight", "model_year"]].iloc[0:7, :].values
print(X)
XTX = X.T.dot(X)
print(XTX)

inverse = np.linalg.inv(XTX)
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])

w = inverse.dot(X.T).dot(y)
print(w.sum())
```
